## Introducción

En 1912, el personal encargado del embarque del Titanic utilizaba listas en papel para registrar a los pasajeros. En esta práctica, estudiaremos ese registro como un ejemplo de procesamiento manual de transacciones. Las listas cumplían una función operativa: dejar constancia de quiénes viajaban. El dataset reúne datos de los pasajeros, como su clase y puerto de embarque, e incorpora un dato conocido después del naufragio: si sobrevivieron.

# **Solución en Python**
Las siguientes celdas responden las diez preguntas del enunciado. Se usa pandas para generar tablas y se incluyen verificaciones para comprobar que los subtotales sumen 891. Los datos originales no se modifican.

```python
# Seleccionar train.csv
from google.colab import files
import pandas as pd
from IPython.display import display

archivos_subidos = files.upload()  # Selecciona únicamente train.csv
nombre_archivo = next(iter(archivos_subidos))
df = pd.read_csv(nombre_archivo)

print('Archivo cargado:', nombre_archivo)
print('Dimensiones del archivo:', df.shape)
display(df.head())

# Comprobar los campos faltantes en todas las columnas
tabla_faltantes = df.isna().sum().rename_axis('Columna').reset_index(name='Valores faltantes')
display(tabla_faltantes)
```

**Pregunta 1.** Pasajeros y PassengerId únicos

```python
total_pasajeros = len(df)
ids_distintos = df['PassengerId'].nunique()

tabla_1 = pd.DataFrame({
    'Indicador': ['Pasajeros registrados', 'PassengerId distintos'],
    'Cantidad': [total_pasajeros, ids_distintos]
})
display(tabla_1)
print('¿Coinciden los dos resultados?:', 'Sí' if total_pasajeros == ids_distintos else 'No')
```

**Pregunta 2.** Pasajeros por clase

```python
tabla_2 = df.groupby('Pclass').size().reset_index(name='Pasajeros')
display(tabla_2)
assert tabla_2['Pasajeros'].sum() == len(df)
clase_mayor = tabla_2.loc[tabla_2['Pasajeros'].idxmax(), 'Pclass']
print('Clase con más pasajeros:', clase_mayor)
```

**Pregunta 3.** Puerto de embarque, incluidos los faltantes

```python
puertos = {'S': 'Southampton', 'C': 'Cherburgo', 'Q': 'Queenstown'}
puerto_nombre = df['Embarked'].map(puertos).fillna('Sin registrar')
tabla_3 = puerto_nombre.value_counts().rename_axis('Puerto').reset_index(name='Pasajeros')
display(tabla_3)
assert tabla_3['Pasajeros'].sum() == len(df)
print('Los puertos faltantes se agruparon como "Sin registrar".')
```

**Pregunta 4.** Tarifa promedio, mínima y máxima por clase

```python
tabla_4 = (df.groupby('Pclass')['Fare']
           .agg(Tarifa_promedio='mean', Tarifa_minima='min', Tarifa_maxima='max')
           .round(2).reset_index())
display(tabla_4)
print('Tarifas sin registrar:', df['Fare'].isna().sum())
print('Pandas omite valores faltantes al calcular promedios, mínimos y máximos.')
```

**Pregunta 5.** Pasajeros de cada clase por puerto

```python
tabla_5 = pd.crosstab(df['Pclass'], df['Embarked'].fillna('Sin registrar'))
tabla_5 = tabla_5.reindex(columns=['C', 'Q', 'S', 'Sin registrar'], fill_value=0)
tabla_5['Total'] = tabla_5.sum(axis=1)
display(tabla_5.reset_index())
assert tabla_5['Total'].sum() == len(df)
print('C = Cherburgo; Q = Queenstown; S = Southampton.')
```

**Pregunta 6.** Sobrevivientes y fallecidos por clase

```python
tabla_6 = pd.crosstab(df['Pclass'], df['Survived'])
tabla_6 = tabla_6.rename(columns={0: 'Fallecieron', 1: 'Sobrevivieron'})
tabla_6 = tabla_6.reindex(columns=['Fallecieron', 'Sobrevivieron'], fill_value=0)
tabla_6['Total'] = tabla_6.sum(axis=1)
display(tabla_6.reset_index())
assert tabla_6['Total'].sum() == len(df)
```

**Pregunta 7.** Pasajeros que viajan con 0, 1 o 2+ familiares

```python
familiares = df['SibSp'] + df['Parch']
grupos = pd.cut(familiares, bins=[-1, 0, 1, float('inf')],
                labels=['0 familiares', '1 familiar', '2 o más familiares'])
tabla_7 = (grupos.value_counts(sort=False)
           .rename_axis('Familiares a bordo').reset_index(name='Pasajeros'))
display(tabla_7)
assert tabla_7['Pasajeros'].sum() == len(df)
print('Se cuenta SibSp + Parch para cada pasajero.')
```

**Pregunta 8.** Rango de edad por clase, incluidos los faltantes

```python
grupo_edad = pd.Series('Sin registrar', index=df.index, dtype='object')
grupo_edad.loc[df['Age'] < 18] = 'Menores de 18'
grupo_edad.loc[df['Age'] >= 18] = '18 años o más'

tabla_8 = pd.crosstab(df['Pclass'], grupo_edad)
tabla_8 = tabla_8.reindex(columns=['Menores de 18', '18 años o más', 'Sin registrar'], fill_value=0)
tabla_8['Total'] = tabla_8.sum(axis=1)
display(tabla_8.reset_index())
assert tabla_8['Total'].sum() == len(df)
print('Las edades faltantes se muestran por separado, sin inventar edades.')
```

**Pregunta 9.** Boletos distintos, repetidos y más frecuentes

```python
conteo_boletos = df['Ticket'].value_counts()
maximo = conteo_boletos.max()
tabla_9 = pd.DataFrame({
    'Indicador': ['Códigos distintos', 'Códigos asociados a más de un pasajero',
                  'Máximo de registros por código'],
    'Cantidad': [df['Ticket'].nunique(), (conteo_boletos > 1).sum(), maximo]
})
display(tabla_9)
print('Códigos que aparecen en más registros:')
top_boletos = conteo_boletos[conteo_boletos == maximo]
display(top_boletos.rename_axis('Código de boleto').reset_index(name='Pasajeros'))
print('Boletos sin registrar:', df['Ticket'].isna().sum())
```

**Pregunta 10.** Cabina registrada o faltante por clase

```python
tabla_10 = pd.crosstab(df['Pclass'], df['Cabin'].notna())
tabla_10 = tabla_10.rename(columns={True: 'Con cabina', False: 'Sin cabina'})
tabla_10 = tabla_10.reindex(columns=['Con cabina', 'Sin cabina'], fill_value=0)
tabla_10['Total'] = tabla_10.sum(axis=1)
display(tabla_10.reset_index())
assert tabla_10['Total'].sum() == len(df)
print('Clase con más registros sin cabina:', tabla_10['Sin cabina'].idxmax())
print('Los registros vacíos en Cabin se cuentan como "Sin cabina".')
```
